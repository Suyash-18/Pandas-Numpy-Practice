### Filter Data based on multiple conditions
#### Useing & and |

```python
#Here We are seeing the companies and number of laptops by them which are expensive and less than 15 inches in size
df_laptops[(df_laptops['Price_tier']=='Expensive') & (df_laptops['Size']=='Chisai')].value_counts('Company')
```

```python
df_laptops[(df_laptops['Price_tier']=='Expensive') & ((df_laptops['Company']=='Apple') | (df_laptops['Company']=='Dell'))].value_counts('Company')
```

```python
#Creating a condition column based on more than two conditions/choices
conditions = [
    df_laptops['Price_euros']>3000.00,
    (df_laptops['Price_euros']>2000.00) & (df_laptops['Price_euros']<=3000.00),
    (df_laptops['Price_euros']>800.00) & (df_laptops['Price_euros']<=2000.00),
    df_laptops['Price_euros']<=800.00
]

values = ['Too Expensive', 'Expensive', 'Affordable', 'Cheap']
```

```python
df_laptops['Price_category']=np.select(conditions,values,default='')
```

```python
df_laptops['Price_category'].value_counts()
```

```python
#Creating a condition column based on more than two conditions/choices
conditions = [
    (df_laptops['Price_euros']>3000.00) & (df_laptops['Size']=='Chisai'),
    (df_laptops['Price_euros']>3000.00) & (df_laptops['Size']=='Dekai'),
    ((df_laptops['Price_euros']>2000.00) & (df_laptops['Price_euros']<=3000.00))& (df_laptops['Size']=='Chisai'),
    ((df_laptops['Price_euros']>2000.00) & (df_laptops['Price_euros']<=3000.00))& (df_laptops['Size']=='Dekai'),
    ((df_laptops['Price_euros']>800.00) & (df_laptops['Price_euros']<=2000.00))& (df_laptops['Size']=='Chisai'),
    ((df_laptops['Price_euros']>800.00) & (df_laptops['Price_euros']<=2000.00))& (df_laptops['Size']=='Dekai'),
    df_laptops['Price_euros']<=800.00
]

values = ['Takai', 'Kangaerareru', 'Kangaerareru','Saiteki','Saiteki','Yasui','Yasui']
```

```python
df_laptops['Actual_value']=np.select(conditions,values,default='')
```
