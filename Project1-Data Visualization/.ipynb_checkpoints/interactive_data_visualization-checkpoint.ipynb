### Here we will use a Plotly library which is going to be used for interactive visuals
This Plotly library is actually the main lib used for visualization of data and express module is used.

```python
import pandas as pd
from IPython.display import display, HTML
```

```python
df=pd.read_csv('population_total.csv')
```

```python
df.dropna(inplace=True)
df_pivot=df.pivot(index='year', columns='country', values='population')
```

```python
df_pivot = df_pivot[['United States', 'United Kingdom','India', 'China', 'Japan', 'South Korea']]
```

```python
df_pivot
```

```python
import plotly.express as px
# df_pivot.iplot(kind='line')
```

```python
# px is new lib its args are df, title is whole graphs title and lables are for x & y axis value is default
px.line(df_pivot, 
        title='Population wise graph',
        labels={'year': 'Years', 'value': 'Population in Billions'}
       )
```

```python
yr2020 = df_pivot[df_pivot.index.isin([2020])]
```

```python
yr2020
```

```python
yr2020=yr2020.T
yr2020
```

```python
# here color=index represint colour based on index of our df so all index have diffrent color
px.bar(yr2020, color=yr2020.index)
```

```python
sample=df_pivot[df_pivot.index.isin([1980,1985,1990,1995,2000,2005,2010,2015,2020])]
```

```python
sample
```

```python
# here barmode is spcified group because we want bars of each country side by side not stacked one above another
px.bar(sample, barmode='group')
```

```python
yr2020.rename(columns={2020:'2020'},inplace=True)
```

```python
# here names should be defined to get the result along with the side sub graph containing names and values is compulsary
px.pie(yr2020, values='2020', names=yr2020.index, title='Population pie in percentage')
```

```python
px.box(df_pivot['India'])
```

```python
px.box(df_pivot,height=800, color='country')
```

```python
px.histogram(df_pivot[['India', 'China']])
```

```python
px.scatter(df_pivot)
```
