### Making Interactive Data Visualization using pivot() and pivot_table()

```python
import pandas as pd
```

```python
df=pd.read_csv('population_total.csv')
```

```python
df
```

### Making a Pivot table

```python
df.dropna(inplace=True)
df_pivot=df.pivot(index='year', columns='country', values='population')
```

```python
df_pivot
```

```python
df_pivot=df_pivot[['United States', 'United Kingdom','India', 'China', 'Japan', 'South Korea']]
```

```python
df_pivot
```

### Using new methord called plot for data visualization

```python
# here kind meaning what type of visualizationdo you want to create here line graph is visualized
df_pivot.plot(kind='line',
              xlabel='Years', ylabel='Population in billion',
              title='Population (1955-2020)',
              figsize=(6,6)
             )
```

```python
# Bar plot
```

```python
yr2020 = df_pivot[df_pivot.index.isin([2020])]
```

```python
yr2020
```

```python
# T here is transpose which mean row be col and col be row
yr2020=yr2020.T
```

```python
yr2020
```

```python
yr2020.plot(kind='bar', 
            xlabel='Countries', ylabel='Population in billion',
            color='orange', title='Population 2020'
           )
```

```python
yr_grp=df_pivot[df_pivot.index.isin([1995, 2000, 2005, 2010, 2015, 2020])]
```

```python
yr_grp
```

```python
yr_grp.plot(kind='bar', xlabel='Year', ylabel='Population', title="Population graph through out the year's")
```

```python
# Pie chart
```

```python
yr2020.rename(columns={2020.0:'2020'}, inplace= True)
```

```python
yr2020
```

```python
yr2020.plot(kind='pie', y='2020', title='Population in 2020(in %)')
```
