### Making Interactive Data Visualization using pivot() and pivot_table()

```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd
```

```python
df=pd.read_csv('population_total.csv')
```

```python
df
```

### Making a Pivot table

```python
df.dropna(inplace=True)
df_pivot=df.pivot(index='year', columns='country', values='population')
```

```python
df_pivot
```

```python
df_pivot=df_pivot[['United States', 'United Kingdom','India', 'China', 'Japan', 'South Korea']]
```

```python
df_pivot
```

### Using new methord called plot for data visualization

```python
# here kind meaning what type of visualizationdo you want to create here line graph is visualized
df_pivot.plot(kind='line',
              xlabel='Years', ylabel='Population in billion',
              title='Population (1955-2020)',
              figsize=(6,6)
             )
# this helps us download the graph visuals in form of image 
plt.savefig("line.png")
# and thus is used to show only graph visual no extra data
plt.show()
```

```python
# this will convert our df to excel file and it will be available at the root directory
df_pivot.to_excel('pivot_tab.xlsx')
```

```python
# Bar plot
```

```python
yr2020 = df_pivot[df_pivot.index.isin([2020])]
```

```python
yr2020
```

```python
# T here is transpose which mean row be col and col be row
yr2020=yr2020.T
```

```python
yr2020
```

```python
yr2020.plot(kind='bar', 
            xlabel='Countries', ylabel='Population in billion',
            color='orange', title='Population 2020'
           )
```

```python
yr_grp=df_pivot[df_pivot.index.isin([1995, 2000, 2005, 2010, 2015, 2020])]
```

```python
yr_grp
```

```python
yr_grp.plot(kind='bar', xlabel='Year', ylabel='Population', title="Population graph through out the year's")
```

```python
# Pie chart here y is to be compulsarily specified
```

```python
yr2020.rename(columns={2020.0:'2020'}, inplace= True)
```

```python
yr2020
```

```python
yr2020.plot(kind='pie', y='2020', title='Population in 2020(in %)')
```

### Boxplot
##### Boxplot is a graph that shows how the data is spread out it shows 5 metrics
- Minimum score
- First Quater
- Median (second quater)
- Third Quater
- Maximum score

```python
df_pivot
```

```python
df_pivot['Japan'].plot(kind='box', ylabel='Population in bajillions', title='Population of india stats')
```

```python
df_pivot.plot(kind='box', ylabel='Population in billions', title='Box Plot of Population')
```

### Histogram
##### histogram is a graph that organize group of data points to ranges
- The ranges are displayed using bars
- These bars show the frequency distribution of data

```python
df_pivot[['Japan', 'South Korea']].plot(kind='hist')
```

### ScatterPlot: It shows the how much x-axis is affected by y-axis
