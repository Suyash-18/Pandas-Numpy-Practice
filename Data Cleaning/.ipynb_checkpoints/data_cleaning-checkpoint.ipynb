```python
# shows all the null values as true and non null values asfalse
df.isnull()
```

```python
# number of all the null values in column
df.isnull().sum().sort_values(ascending=False)
```

```python
for columns in df.columns:
    p = df[columns].isnull().mean()*100
    print(columns + ' : ' + str(round(p,2)))
```

```python
# we can do 3 things delete whole column where the null values are
# df.drop('director', axis=1)

# we can del row in which the values are null for a specific column
# delete=df[df['director'].isnull()].index
# df.drop(delete, axis=0)

# delete rows without drop
# df[~(df['director'].isnull())]

#delete row using dropna
df.dropna(subset=['director'])
```

```python
df.isnull().sum()
```

```python
df[df['rating'].isnull()]
```

```python
# this will fill the values of the categorial type data not applicable for normal strings 
# mean , median can be used for numeric values
mode=''.join(df['rating'].mode())
df.fillna({'rating':mode},inplace=True)
df.iloc[6827]
```

```python
df[df['duration'].isnull()]
```

```python
# all the null values in duration is filled by 0 so nan is replaced by 0 or any data of our liking
df['duration'].fillna("0").isnull().sum()
```

```python
# this will make next value in the column to be in place of null
df.bfill()
```

```python
# this will fill null value with the value of the same column before null value
df.ffill()
```

```python
# Extraction of useful data using split() and extract() methords
dfm=df[df['type']=='Movie']
```

```python
dfm
```

```python
dfm['minutes']=dfm['duration'].str.split(expand=True)[0].astype(int)
```

```python
dfm
```

```python
# here ',' means seprate when , comes in the string
dfm['date_added'].str.split(',')
```

```python
# in extract we use regex to extract the data here we extract 4 continous digits meaning year
dfm['date_added'].str.extract(r'(\d{4})')
```

### Identify outlires

```python
# histogram
eg=dfm['minutes']
px.histogram(eg, nbins=10)
```

```python
# value_count here lets us know the number of present count accoring to numbers of bins
#by seing this data we can select to only move ahead with normal data not outlires
# here opu lires are movies above 218 and below 31
eg.value_counts(bins=10)
```

```python
# no more outlires
dfm[((dfm['minutes'] > 31) & (dfm['minutes'] < 218))]
```

```python
# here we gets all the outlires
dfm[~((dfm['minutes'] > 31) & (dfm['minutes'] < 218))]
```

```python
px.box(eg, orientation='h')
```

```python
# we can get the lower fence and upper fence using formulas as well
# IQR=Q3(75%)-Q1(25%)
# LF= Q1-1.5*(Q3-Q1)
# UF= Q3+1.5*(Q3-Q1)
# WE CAN GET Q3 AND Q1 using describe() methord
dfm['minutes'].describe()
```

```python
iqr=114-87
lf=87-1.5*iqr
uf=114+1.5*iqr
print('lower fence:',lf)
print('upper fence:',uf)
```

```python
# now we know fences so we can remove outlires based on fence outcome as well
dfm[~((dfm['minutes']>46) & (dfm['minutes']<154))]
```

```python
# bar plot to display categorical data
px.bar(dfm['rating'].value_counts())
```
