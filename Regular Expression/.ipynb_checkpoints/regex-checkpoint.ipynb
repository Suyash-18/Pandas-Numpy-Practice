```python
import re
```

```python
text="""Hello World.
I love Python!!!!!
My first lucky number is 987-654-321
My second lucky number is 876-543-210
My third lucky number is 765-432-100
My fourth lucky number is 123-456-123-123-456"""
```

```python
text
```

```python
# here r'' tells code that inside '' is regex code 
# flags re.I means case insencitive meaning no upper or lower case to be concider while search
re.search(r'^hello', text, flags=re.I)
```

```python
re.findall(r'\d', text)
```

```python
re.findall(r'[^\w\s]', text)
```

```python
text = '''
13-04-2021
2021-13-04
2021-04-13
'''
```

```python
re.findall(r'\d{2}-\d{2}-\d{4}', text)
```

```python
text = '''
Username10
abc
10
'''
```

```python
re.findall(r'[a-z0-9]{4,14}', text, flags=re.I)
```
