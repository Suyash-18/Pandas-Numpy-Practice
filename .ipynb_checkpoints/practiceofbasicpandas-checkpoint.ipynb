### This is the project of Web Scraping With pandas
this is the site we are scraping: https://football-data.co.uk/data.php

```python
import pandas as pd
```

```python
df_premier21 = pd.read_csv('https://football-data.co.uk/mmz4281/2122/E0.csv')
```

```python
df_premier21
```

```python
df_premier21.rename(columns={'FTHG':'home_goals', 'FTAG':'away_goals'}, inplace=True)
```

```python
df_premier21
```

```python
leauges = ['E0', 'E1', 'E2', 'E3', 'EC']
frames = []
for a in leauges:
    df=pd.read_csv('https://football-data.co.uk/mmz4281/2122/'+ a +'.csv', encoding='unicode_escape')
    frames.append(df)
```

```python
len(frames)
```

```python
frames[4]
```

```python
all_frames = []

for a in leauges:
    for b in range(15,21):
        year=str(b)+str(b+1)
        df=pd.read_csv('https://football-data.co.uk/mmz4281/'+year+'/'+ a +'.csv', encoding='unicode_escape')
        df.insert(1, 'season', f"Season {b}")
        all_frames.append(df)
```

```python
len(all_frames)
```
