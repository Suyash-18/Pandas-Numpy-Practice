```python
#using assign() to add multiple columns
score1 = np.random.randint(1, 100, size=1000)
score2 = np.random.randint(1, 100, size=1000)
```

```python
series1 = pd.Series(score1, index=np.arange(0, 1000))
series2 = pd.Series(score1, index=np.arange(0, 1000))
```

```python
#This is only a copy orignal Dataframe is not changed
df_exams.assign(score1=series1, score2=series2)
```

```python
#Dataframe reflects changes when we assign using = operator
df_exams = df_exams.assign(score1=series1, score2=series2)
```

```python
df_exams
```

```python
#Adding new column using insert() this adds into orignal Dataframe here 1 is index in columns test is name and series1 is values
df_exams.insert(1, 'test', series1)
```
