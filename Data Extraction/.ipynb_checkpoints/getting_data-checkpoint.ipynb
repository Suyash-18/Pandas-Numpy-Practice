## Data Extraction using pandas
#### loc[row_lable,column_lable] and iloc[row_position,column_position] will be used

```python
import pandas as pd
```

```python
df = pd.read_csv('players_20.csv')
```

```python
# Set index to names
df.set_index('short_name', inplace=True)
```

```python
df
```

```python
df=df[['long_name','age','dob','height_cm','weight_kg','nationality','club']]
```

```python
df
```

```python
df.loc['L. Messi']
```

```python
df.loc['L. Messi', 'height_cm']
```

```python
# Here : is representing all row_lables/indexes
df.loc[:, 'height_cm']
```

```python
df.loc['L. Messi', :]
```

```python
df.loc[['L. Messi', 'Cristiano Ronaldo']]
```

```python
df.loc[['L. Messi', 'Cristiano Ronaldo'], 'height_cm']
```

```python
df.loc['L. Messi',[ 'height_cm', 'weight_kg']]
```

```python
df.loc[['L. Messi', 'Cristiano Ronaldo'],[ 'height_cm', 'weight_kg']]
```

```python
# Using : and defining we want data from a:b a to b
df.loc[['L. Messi', 'Cristiano Ronaldo'], 'age':'club']
```

```python
df.index[:10]
```

```python
df.loc['L. Messi':'M. Salah',['age','dob','height_cm','weight_kg','nationality','club']]
```

```python
df.loc['L. Messi':'M. Salah','age':'club']
```

```python
df.loc[df['height_cm']>180, 'age':'club']
```

```python
df.loc[(df['nationality']=='Argentina') & (df['height_cm']>180), 'age':'club'].head(10)
```

```python
# Selecting elements by using iloc[]
df.iloc[0]
```

```python
df.iloc[0,3]
```

```python
df.iloc[1,4]
```

```python
df.iloc[:,3]
```

```python
df.iloc[0,:]
```

```python
df.iloc[[0,1],:]
```

```python
df.iloc[[0,1],3]
```

```python
df.iloc[0,[3,4]]
```

```python
df.iloc[[0,1],[3,4]]
```

```python
# Here value after :7 is excluded
df.iloc[[0,1],1:7]
```

```python
df.iloc[0:10, 1:7]
```

```python
# iloc[] dose not take the df series as a parameter so we need to convert it into a list to give it to iloc[]
df.iloc[list(df['height_cm']>180), 1:5]
```

```python
df.iloc[list(df['height_cm']>180), 1:7].drop_duplicates('height_cm').sort_values('height_cm')
```

```python
df.iloc[list((df['height_cm']>180) & (df['nationality']=='England')) , 1:7]
```
