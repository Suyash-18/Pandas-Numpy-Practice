### Sample Data Generation

```python
import pandas as pd
```

```python
df = pd.read_csv('players_20.csv')
```

```python
df.set_index('short_name', inplace=True)
```

```python
df=df[['long_name','age','dob','height_cm','weight_kg','nationality','club']]
df
```

```python
# Sample(n) methord is used to extract n number of random elements 
# here random_state=99 gets us exact same n elements every time raandom_state is used to controle randomness
df['nationality'].sample(10, random_state=99)
```

```python
# extract a certain percentage (2% here) of data frame using frac(fraction)
df.sample(frac=0.02, random_state=12)
```

```python
# we can increase the sample by making frac>1 amd also write replace=True
df.sample(frac=1.234, replace=True)
```

### Filter Datafram using Query() methord

```python
# in query methord we need to write the condition in ''
df.query('age>34').head(5)
```

```python
# this was our old filter methord it is also known as boolean slicing
df[df['age']>34].head(5)
```

```python
# in query methord "" inside of '' is used for the purpose of diffrentiating between two conditions
# instaed of & we use "and" here
df.query('age>34 and nationality=="Italy"').head(5)
```

```python
df[(df['age']>34) & (df['nationality']=="Italy")].head(5)
```

```python
# not() operator is used to do the work of ~(!) 
df.query('not(age>34)').head(5)
```

```python
df[df['age']>~34].head()
```

```python
df.query('height_cm/100 > 1.8').head()
```

```python
df[(df['height_cm']/100)>1.8].head()
```

```python
df.dtypes
```

```python
df = df.copy()
df['dob'] = df['dob'].astype('datetime64[ns]')
```

```python
df.dtypes
```

```python
df[df['dob'].dt.year > 1990]
```

```python
# using query
df.query("dob.dt.year > 1990")
```

### Apply() methord used to apply functions of numpy

```python
# Numpy functionality is introduced inside the ()
import numpy as np
df['age'].apply(np.sqrt)
```

```python
def calculate_bmi(datafram):
    return datafram['weight_kg']/(datafram['height_cm']/100)**2
```

```python
df.apply(calculate_bmi, axis=1)
```

## Lambda Function

```python
# first lets see basic function
def sum_v(a,b):
    x= a+b
    return x
```

```python
sum_v(2,4)
```

```python
# now Lambda function
sum_lambda = lambda a,b:a+b
```

```python
sum_lambda(3,4)
```

```python
# We can use lambda func without a name here because we are only going to use it once
df['height_cm'].apply(lambda x:x/100)
```

```python
df['long_name'].apply(lambda x:x.upper())
```

```python
df['dob'].apply(lambda x:x.year)
```

```python
df.apply(lambda x: x['weight_kg']/((x['height_cm']/100)**2), axis=1)
```

### Copy of a dataframe using copy() methord

```python
# in copy we have deep=True as default parameter
# this will allow the copy to be independent copy not just a holder pointing to orignal
dfc=df.copy()
```

```python
df.loc['L. Messi', 'height_cm'] = 200
```

```python
df
```

```python
dfc
```

```python
# by setting deep to false we can say that the new dfcopy is just pointing to orignal df it is not a new entity
dfcopy = df.copy(deep=False)
dfcopy
```

```python
df.loc['L. Messi', 'height_cm']=180
```

```python
df
```

```python
dfcopy
```

```python
# We can also make copy by just assigning it with = operator it will be shallow meaning deep=False
df_new=df
```
