```python
import pandas as pd
```

```python
df1=pd.DataFrame({'id':['A','B','C','D'],
              'age': [25,27,24,22]
             })
df2=pd.DataFrame({'id':['C','D','E','F'],
                  'job':['Accountant', 'Developer','Founder','Unemployed']
                 })
```

```python
df1
```

```python
df2
```

```python
# sample here shows inner joins overlaping based on ids
df1.merge(df2, on='id', how='inner')
```

```python
df_m=pd.read_csv('IMDb movies.csv', low_memory=False)
df_r=pd.read_csv('IMDb ratings.csv')
```

```python
df_m= df_m[['imdb_title_id', 'title', 'year', 'genre', 'country', ]]
df_r=df_r[['imdb_title_id', 'total_votes', 'mean_vote']]
```

```python
# inner is default
df_m.merge(df_r, on='imdb_title_id')
```

```python
# outer just merge it all
df1.merge(df2, on='id', how='outer')
```

```python
df_m.merge(df_r, on='imdb_title_id', how='outer')
```

```python
df1.merge(df2,on='id',how='outer', indicator=True)
```

```python
# Exclusive full join all intersecting values are dropped
df1.merge(df2,on='id',how='outer', indicator=True).query('_merge=="left_only" or _merge=="right_only"')
```

```python
df_m.merge(df_r,on='imdb_title_id', how='outer', indicator=True)
```

```python
df_m.merge(df_r,on='imdb_title_id', how='outer', indicator=True).query('_merge=="left_only" or _merge=="right_only"')
```

```python
df1.merge(df2, on='id',how='left')
```

```python
sam=df_m.sample(frac=0.5)
```

```python
mersam=sam.merge( df_r, on='imdb_title_id', how='left')
```

```python
mersam.shape
```

```python
mersam
```

```python
# Outer is set to how only when exclusive full left or right is needed
df1.merge(df2,on='id', how='outer', indicator=True).query('_merge=="left_only"')
```
